```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None        # address of next memory block of Node type
    def __repr__(self):
        return f"Node({self.data})"

class LinkedList:
    def __init__(self, items=None):
        self.head = None
        self.tail = None
        self.size = 0

        if items is not None:
            for item in items:
                self.append(item)
    
    def __str__(self):
        # Optimal: O(N) to build the string representation as it traverses the list.
        # user-friendly string representation of the list.
        # e.g., "A -> B -> C -> None"
        nodes_str = []
        current = self.head
        while current:
            nodes_str.append(str(current.data))
            current = current.next
        return " -> ".join(nodes_str) + " -> None"

    def __repr__(self):
        # Optimal: O(N) to build the representation.
        # Developer-friendly string representation.
        # e.g., LinkedList([A, B, C])
        nodes_data = []
        current = self.head
        while current:
            nodes_data.append(repr(current.data))
            current = current.next
        return f"LinkedList([{', '.join(nodes_data)}])"
    
    def __len__(self):
        return self.size

    def __iter__(self):
        # Optimal: O(1) to return the iterator object. Traversal is O(N).
        current = self.head
        while current:
            yield current.data
            current = current.next
        
    def is_empty(self):
        # Optimal: O(1) to check if the list is empty.
        return self.size == 0

    def append(self, data):
        # Optimal: O(1) to append at the end of the list.
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            # # without tail optimization, this would be O(N) as we would need to traverse the list.
            # current = self.head
            # while current.next:
            #     current = current.next
            # current.next = new_node
            
            # If tail optimization is used, we can directly append to the tail.
            self.tail.next = new_node
            self.tail = new_node
        self.size += 1

    def prepend(self, data):
        # Optimal: O(1) to prepend at the start of the list.
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head = new_node
        self.size += 1

    def insert(self, index, data):
        # Optimal: O(N) in the worst case (inserting at the end or near end without tail optimization for this specific op).
        # O(N) - inserting somewhere near the middle point of the Linked List.
        # O(1) for inserting at the beginning (index 0).
        # Inserts data at a specific index.

        if not (0 <= index <= self.size):       # self.size is allowed - reduces to append
            raise IndexError("Index out of bounds for insert operation.")
        if index == 0:
            self.prepend(data)
        elif index == self.size:
            self.append(data)
        else:
            new_node = Node(data)
            current = self.head
            # suppose - want to insert at index 3 (0,1,2,3) - We need pointer at index = 2(0,1,2), just 1 unit before the insert point.
            # initially at head - index = 0, 0-base indexing.

            for _ in range(index):  # excluding index ->
                current = current.next
            new_node.next = current.next
            current.next = new_node
            self.size += 1

    def compare(self, a, b) -> bool:
        # Optimal: O(1) for basic data types (int, str, etc.). O(N) for complex objects.
        # Compares two data items for equality.
        return a == b
    
    def search(self, data) -> bool:
        # Optimal: O(N) in the worst case (item is at the end or not present).
        # Searches for the first occurrence of data. Returns True if found, else False.
        current = self.head
        while current:
            if self.compare(current.data, data):
                return True
            current = current.next
        return False

    def searchRecursive(self, targetData, currentNode) -> bool:
        # Optimal: O(N) in the worst case (item is at the end or not present).
        # Searches for the first occurrence of data. Returns True if found, else False.
        if not currentNode:
            return False
        if currentNode.data == targetData:
            return True
        return self.searchRecursive(targetData, currentNode.next)

    def getAllOccurrences(self, targetData, currentNode, index: int = 0, resultIndexList: list = []) -> list:
        # Optimal: O(N) in the worst case (item is at the end or not present).
        # Searches for all occurrences of data and returns a list of indices where it occurs.
        # RECURSIVE
        if not currentNode:
            return resultIndexList
        if currentNode.data == targetData:
            resultIndexList.append(index)
        return self.getAllOccurrences(targetData, currentNode.next, index+1, resultIndexList)
    
    def get(self, index):
        # Optimal: O(N) in the worst case. O(1) for index 0.
        # Retrieves data at a specific index.
        if not(0 <= index <= self.size - 1):
            raise IndexError("Index out of Bounds for Get Operation.")
        
        current = self.head
        for _ in range(index):
            current = current.next
        return current.data

    def delete_by_value(self, data):
        # Optimal: O(N) in the worst case. O(1) if deleting the head.
        # Deletes the first node containing the specified data.

        # 3 cases - 0 nodes, 1 node, more than 1 node.
        if self.is_empty():
            raise ValueError("Linked List is empty, deletion not possible")
        if self.head.data == data:
            temp = self.head
            self.head = self.head.next
            if self.head is None:
                self.tail = None
            del temp
            self.size -= 1 

        else:
            pass
    
  
```
