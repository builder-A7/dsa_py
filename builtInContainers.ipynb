```python
l1 = list()

l1.append(1)
l1.append(2)
l1.append(3)
l1.append(2)

print(l1)

print("COUNT of element {} = {}".format(1, l1.count(2)))
print(l1.index(2))

print("SUM = ", sum(l1))
print("MAX = ", max(l1))
print("MIN = ", min(l1))
print("LENGTH = ", len(l1))

print("REVERSED = ", list(reversed(l1)))
print("Did original list get reversed? NO", l1)

print("SORTED = ", sorted(l1))
print("Did original list get sorted? NO", l1)

l1.sort()
print("AFTER SORT - in Place sort of the actual list - original list modified = ", l1)

print("POP = ", l1.pop())
print("AFTER POP = removal of the last element", l1)

# list_reference.insert(index, element)
l1.insert(5, 5)
print("AFTER INSERT = ", l1)
# initially - [1,2,2,3] - Then, pop() removes 3 - Now, [1,2,2] - Then, insert(5,5) inserts 5 at index 5 - Now, [1,2,2,5].
# But still, only 4 elements.
print(len(l1))

print(l1.index(5))  # This will give the index of the first occurrence of 5. I tried to insert at index 5 directly, after index 2.
print("Index of 5 = ", l1.index(5))     # But it still got inserted at index 3, as there were only 3 elements in the list initially.

# print     # O/p of print -=> <function print(*args, sep=' ', end='\n', file=None, flush=False)>
```

```python
t1 = tuple()
t1 = (1, 2, 3, 4, 5)
print("Tuple t1 = ", t1)
print("Tuple element at 0th index = ", t1[0])

print("Count of element {} in tuple 1 = {}".format(1, t1.count(1)))
print("Index of element 2 = ", t1.index(2))  # Index of element 2 in tuple t1

# t1.__add__(1)  # This will not modify the tuple, as tuples are immutable.
print("Tuple after __add__ = ", t1)  # Still the same tuple, as tuples are immutable.
print("Tuple after __add__ = ", t1 + (1,))  # This will return a new tuple, but not modify the original tuple.
print("Tuple after __add__ = ", t1 + (6, 7, 8))  # This creates a new Tuple with multiple elements added, original tuple unmodified.
print("t1 still remains the same = ", t1)  # Original tuple remains unchanged.
```
