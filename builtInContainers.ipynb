# List operations

```python
l1 = list()

l1.append(1)
l1.append(2)
l1.append(3)
l1.append(2)

print(l1)

print("COUNT of element {} = {}".format(1, l1.count(2)))
print(l1.index(2))

print("SUM = ", sum(l1))
print("MAX = ", max(l1))
print("MIN = ", min(l1))
print("LENGTH = ", len(l1))

print("REVERSED = ", list(reversed(l1)))
print("Did original list get reversed? NO", l1)

print("SORTED = ", sorted(l1))
print("Did original list get sorted? NO", l1)

l1.sort()
print("AFTER SORT - in Place sort of the actual list - original list modified = ", l1)

print("POP = ", l1.pop())
print("AFTER POP = removal of the last element", l1)

# list_reference.insert(index, element)
l1.insert(5, 5)
print("AFTER INSERT = ", l1)
# initially - [1,2,2,3] - Then, pop() removes 3 - Now, [1,2,2] - Then, insert(5,5) inserts 5 at index 5 - Now, [1,2,2,5].
# But still, only 4 elements.
print(len(l1))

print(l1.index(5))  # This will give the index of the first occurrence of 5. I tried to insert at index 5 directly, after index 2.
print("Index of 5 = ", l1.index(5))     # But it still got inserted at index 3, as there were only 3 elements in the list initially.

# print     # O/p of print -=> <function print(*args, sep=' ', end='\n', file=None, flush=False)>
```

# Tuple operations

```python
t1 = tuple()
t1 = (1, 2, 3, 4, 5)
print("Tuple t1 = ", t1)
print("Tuple element at 0th index = ", t1[0])
print("Length of Tuple t1 = ", len(t1))

print("Count of element {} in tuple 1 = {}".format(1, t1.count(1)))
print("Index of element 2 = ", t1.index(2))  # Index of element 2 in tuple t1

# t1.__add__(1)  # This will not modify the tuple, as tuples are immutable.
print("Tuple after __add__ = ", t1)  # Still the same tuple, as tuples are immutable.
print("Tuple after __add__ = ", t1 + (1,))  # This will return a new tuple, but not modify the original tuple.
print("Tuple after __add__ = ", t1 + (6, 7, 8))  # This creates a new Tuple with multiple elements added, original tuple unmodified.
print("t1 still remains the same = ", t1)  # Original tuple remains unchanged.
```

# Dict Operations

```python
# 1. Creating Dictionaries
empty_dict = {}
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
another_dict = dict(name="Bob", age=25, city="London")
dict_from_tuples = dict([("fruit", "apple"), ("color", "red")])
print("Dictionary contents:", my_dict)

# 2. Accessing Items (by key)
name = my_dict["name"]
print("Value of Name field = ", name)
# age = my_dict["non_existent_key"]  # Raises KeyError if key doesn't exist

# 3. Adding / Updating Items
my_dict["age"] = 31                 # Update value mapped to an existent key
my_dict["occupation"] = "Engineer"  # Add new key-value pair if does not exist already
my_dict.update({"city": "San Francisco", "country": "USA"})
my_dict.update(zipcode=94107, status="active")
print("Updated Dictionary contents:", my_dict)

# 4. Removing Items
# Using del
del my_dict["city"]
# del my_dict["non_existent_key"] # Raises KeyError

# Using pop() - removes specified key and returns the value
country = my_dict.pop("country")
# print(country)  # Output: USA
# If the key does not exist, pop() can raise KeyError unless a default value is provided

# status = my_dict.pop("non_existent_key") # Raises KeyError
status_default = my_dict.pop("non_existent_key", "default_value_if_not_found")

# Using popitem() - removes and returns an arbitrary (key, value) pair (LIFO since Python 3.7)
last_item = my_dict.popitem()
# print(last_item)  # Output: ('status', 'active') or similar, depending on the last item

# 5. Checking for Key Existence
key_exists = "name" in my_dict
key_not_exists = "address" not in my_dict

# 6. Getting Length (number of key-value pairs)
num_items = len(my_dict)

# 7. Iterating
# Iterate over keys (default iteration)
for key in my_dict:
    pass # print(key, my_dict[key])

# Iterate over keys explicitly
for key in my_dict.keys():
    pass # print(key)

# Iterate over values
for value in my_dict.values():
    pass # print(value)

# Iterate over key-value pairs (items)
for key, value in my_dict.items():
    pass # print(key, value)

# 8. Getting Views (dynamic views of dictionary entries)
keys_view = my_dict.keys()
values_view = my_dict.values()
items_view = my_dict.items()

# 9. Copying Dictionaries
# Shallow copy
shallow_copy1 = my_dict.copy()
shallow_copy2 = dict(my_dict)

# For deep copy (if dictionary contains mutable objects like lists):
# import copy
# deep_copy = copy.deepcopy(my_dict)

# 10. Clearing a Dictionary (removes all items)
my_dict.clear()

# 11. Getting a Value with a Default (avoids KeyError)
# Re-populate for example
my_dict = {"name": "Alice", "age": 30}
city = my_dict.get("city")  # Returns None if "city" not found
city_default = my_dict.get("city", "Unknown") # Returns "Unknown" if "city" not found
name_val = my_dict.get("name", "Default Name") # Returns "Alice"

# 12. Merging Dictionaries (Python 3.9+)
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}
merged_dict_pipe = dict1 | dict2 # dict1 values for common keys are overwritten by dict2
# dict1 remains {"a": 1, "b": 2}
# merged_dict_pipe will be {'a': 1, 'b': 3, 'c': 4}

# In-place merge using pipe operator (Python 3.9+)
# dict1 |= dict2
# Now dict1 would be {'a': 1, 'b': 3, 'c': 4}

# Merging using unpacking (Python 3.5+)
merged_dict_unpack = {**dict1, **dict2}
# merged_dict_unpack will be {'a': 1, 'b': 3, 'c': 4}

# Using update() method (works in all Python 3 versions)
# dict1.update(dict2)
# Now dict1 would be {'a': 1, 'b': 3, 'c': 4}

# 13. Set Default (setdefault)
# If key is in the dictionary, return its value.
# If not, insert key with a value of default and return default.
# Re-populate for example
my_dict = {"name": "Alice"}
age = my_dict.setdefault("age", 0) # "age" not in my_dict, so it's added: my_dict becomes {"name": "Alice", "age": 0}, age becomes 0
name = my_dict.setdefault("name", "Bob") # "name" is in my_dict, its value "Alice" is returned, my_dict is unchanged. name becomes "Alice"
```
